# Moving Average (Média Móvel)

## Importação de bibliotecas

```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

## Importação de dataset

```python
data = pd.read_csv("/content/carne-brasileira-exportada.csv", encoding='utf-8', sep=',')
data.head()
```

## Separação de dataset por carne

```python
bovino = data[data['tipo_carne'] == 'bovino'].copy()
bovino.head()
```

```python
suino = data[data['tipo_carne'] == 'suino'].copy()
suino.head()
```

```python
frango = data[data['tipo_carne'] == 'frango'].copy()
frango.head()
```

### Ordenação temporal

```python
bovino = bovino.sort_values(by=['ano', 'trimestre']).reset_index(drop=True)
```

```python
suino = suino.sort_values(by=['ano', 'trimestre']).reset_index(drop=True)
```

```python
frango = frango.sort_values(by=['ano', 'trimestre']).reset_index(drop=True)
```

## Baseline para bovino

### Simple Moving Average (SMA)

```python
sma = bovino['exportacao_usd'].rolling(window=4).mean()
print(sma)
```

```python
y_true = bovino['exportacao_usd'].iloc[3:]
y_pred = sma.iloc[3:]
mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
print(mae, rmse)
```

### Exponential Moving Average (EMA)

```python
ema = bovino['exportacao_usd'].ewm(span=4, adjust=False).mean()
print(ema)
```

```python
y_true = bovino['exportacao_usd'].iloc[3:]
y_pred = ema.iloc[3:]
mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
print(mae, rmse)
```

# Treinamento e Avaliação de Modelo LSTM
Este notebook cria um fluxo completo para treinar e avaliar um modelo LSTM em séries temporais de produção/exportação de carnes brasileiras.
Usaremos as métricas: MSE, RMSE e R².

```python
# =============================================================================
# Configuração Inicial
# =============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.initializers import Constant
import warnings
warnings.filterwarnings('ignore')

# Caminho do arquivo (ajuste se necessário)
PATH = "/content/carne-brasileira-exportada.csv"
df = pd.read_csv(PATH)

# Coluna alvo
target_col = "preco_medio_ton_usd"

# Colunas que serão usadas como features (todas exceto metadados e alvo)
feature_cols = ["producao_cabecas_anual_mi", "abate_cabecas", "peso_carcaca_ton",
                "exportacao_usd", "media_cambio_usdbrl", "custo_milho_rs", "custo_soja_rs"]

# Parâmetros
sequence_length = 4      # 3 anos de dados trimestrais
test_size = 0.2
n_epochs = 80
batch_size = 4

# =============================================================================
# Função para preparar dados de um tipo de carne
# =============================================================================
def prepare_data(df, meat_type, feature_cols, target_col, sequence_length, test_size):
    # Filtra e ordena por data
    meat_df = df[df["tipo_carne"] == meat_type].copy()
    meat_df["date"] = pd.to_datetime(meat_df["ano"].astype(str) + "Q" + meat_df["trimestre"].astype(str))
    meat_df = meat_df.sort_values("date").reset_index(drop=True)
    
    # Seleciona features e target
    X = meat_df[feature_cols].values.astype(float)
    y = meat_df[target_col].values.astype(float).reshape(-1, 1)
    
    # Normalização (features e target separadamente)
    scaler_X = MinMaxScaler(feature_range=(0, 1))
    scaler_y = MinMaxScaler(feature_range=(0, 1))
    X_scaled = scaler_X.fit_transform(X)
    y_scaled = scaler_y.fit_transform(y)
    
    # Divisão treino/teste (respeitando ordem temporal)
    split_idx = int(len(X_scaled) * (1 - test_size))
    X_train, X_test = X_scaled[:split_idx], X_scaled[split_idx:]
    y_train, y_test = y_scaled[:split_idx], y_scaled[split_idx:]
    
    # Criar sequências para o LSTM
    def create_sequences(X, y, seq_len):
        X_seq, y_seq = [], []
        for i in range(len(X) - seq_len):
            X_seq.append(X[i:i+seq_len])
            y_seq.append(y[i+seq_len])
        return np.array(X_seq), np.array(y_seq)
    
    X_train_seq, y_train_seq = create_sequences(X_train, y_train, sequence_length)
    X_test_seq, y_test_seq = create_sequences(X_test, y_test, sequence_length)
    
    return (X_train_seq, y_train_seq), (X_test_seq, y_test_seq), scaler_y, meat_df, split_idx

# =============================================================================
# Modelo LSTM otimizado segundo recomendações do artigo
# =============================================================================
def build_lstm_model(input_shape):
    model = Sequential()
    # Bias inicial negativo para portas de entrada (evita abuse problem)
    model.add(LSTM(
        units=8,                       # poucas unidades devido ao tamanho reduzido dos dados
        activation='tanh',             # h() com range [-1,1]
        recurrent_activation='sigmoid',# sigmoid padrão para portas [0,1]
        bias_initializer=Constant(-2.0), # bias negativo nas portas (input gate e output gate)
        recurrent_dropout=0.2,         # dropout recorrente para regularização
        input_shape=input_shape,
        return_sequences=False
    ))
    model.add(Dropout(0.2))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse', metrics=['mse'])
    return model

# =============================================================================
# Função para treinar e avaliar um tipo de carne
# =============================================================================
def train_evaluate(meat_type):
    print(f"\n{'='*50}")
    print(f"Processando: {meat_type.upper()}")
    print(f"{'='*50}")
    
    # Preparar dados
    (X_train, y_train), (X_test, y_test), scaler_y, meat_df, split_idx = prepare_data(
        df, meat_type, feature_cols, target_col, sequence_length, test_size
    )
    
    if len(X_train) == 0 or len(X_test) == 0:
        print(f"  -> Dados insuficientes para {meat_type} (seq_len={sequence_length})")
        return None
    
    print(f"  Treino: {X_train.shape[0]} sequências | Teste: {X_test.shape[0]} sequências")
    
    # Construir modelo
    model = build_lstm_model((X_train.shape[1], X_train.shape[2]))
    model.summary()
    
    # Early stopping
    early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    
    # Treinamento
    history = model.fit(
        X_train, y_train,
        epochs=n_epochs,
        batch_size=batch_size,
        validation_split=0.2,
        callbacks=[early_stop],
        verbose=1
    )
    
    # Previsões (rolling)
    predictions_scaled = model.predict(X_test)
    predictions = scaler_y.inverse_transform(predictions_scaled)
    actuals = scaler_y.inverse_transform(y_test)
    
    # Métricas
    mse = mean_squared_error(actuals, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(actuals, predictions)
    
    print(f"\n  Resultados para {meat_type}:")
    print(f"  MSE : {mse:.4f}")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  R²  : {r2:.4f}")
    
    # Plot comparativo
    # Ajuste das datas para as previsões
    train_seq_len = len(meat_df[:split_idx]) - sequence_length
    dates_test = meat_df.iloc[split_idx + sequence_length:]['date'].reset_index(drop=True)
    
    plt.figure(figsize=(12, 5))
    plt.plot(dates_test, actuals.flatten(), label='Real', marker='o', linestyle='-')
    plt.plot(dates_test, predictions.flatten(), label='Previsto (LSTM)', marker='x', linestyle='--')
    plt.title(f'Previsão do Preço Médio (USD/ton) - {meat_type.upper()}')
    plt.xlabel('Data')
    plt.ylabel('Preço Médio (USD/ton)')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    return {
        'meat_type': meat_type,
        'mse': mse,
        'rmse': rmse,
        'r2': r2,
        'train_samples': len(X_train),
        'test_samples': len(X_test)
    }

# =============================================================================
# Execução principal
# =============================================================================
meat_types = sorted(df['tipo_carne'].dropna().unique())
results = []

for meat in meat_types:
    res = train_evaluate(meat)
    if res:
        results.append(res)

# Tabela resumo
results_df = pd.DataFrame(results)
print("\n" + "="*60)
print("RESUMO FINAL DOS RESULTADOS")
print("="*60)
print(results_df.to_string(index=False))

# Opcional: salvar resultados
# results_df.to_csv('resultados_lstm_otimizado.csv', index=False)
```
